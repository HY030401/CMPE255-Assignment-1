## Application Design

The application stores tasks in a simple Python list.

Each task contains:

- Task ID
- Task description
- Status

A task can have one of two statuses:

- Pending
- Completed

The user interface communicates with Python functions that update the task list.

```python
tasks = []

next_task_id = 1

print("Task workspace initialized.")
```

```python
def get_task_table():

    if len(tasks) == 0:
        return pd.DataFrame(
            columns=[
                "Task ID",
                "Task",
                "Status"
            ]
        )

    return pd.DataFrame(tasks)
```

```python
def add_task(task_text):

    global next_task_id

    task_text = task_text.strip()

    if task_text == "":
        return (
            get_task_table(),
            "Please enter a task."
        )

    new_task = {
        "Task ID": next_task_id,
        "Task": task_text,
        "Status": "Pending"
    }

    tasks.append(new_task)

    next_task_id += 1

    return (
        get_task_table(),
        "Task added successfully."
    )
```

```python
def complete_task(task_id):

    try:
        task_id = int(task_id)

    except:
        return (
            get_task_table(),
            "Please enter a valid Task ID."
        )

    for task in tasks:

        if task["Task ID"] == task_id:

            task["Status"] = "Completed"

            return (
                get_task_table(),
                f"Task {task_id} completed."
            )

    return (
        get_task_table(),
        "Task ID not found."
    )
```

```python
def delete_task(task_id):

    try:
        task_id = int(task_id)

    except:
        return (
            get_task_table(),
            "Please enter a valid Task ID."
        )

    for task in tasks:

        if task["Task ID"] == task_id:

            tasks.remove(task)

            return (
                get_task_table(),
                f"Task {task_id} deleted."
            )

    return (
        get_task_table(),
        "Task ID not found."
    )
```

```python
add_task("Finish CMPE 255 assignment")
add_task("Review data science notebooks")
add_task("Record YouTube walkthrough")

get_task_table()
```

```python
complete_task(2)

get_task_table()
```

## Interactive Todo Workspace

The following Gradio interface provides an interactive web interface for the task management functions.

Users can add tasks, mark tasks as completed, delete tasks, and view the current workspace.

```python
with gr.Blocks(title="Dynamic Todo Workspace") as demo:

    gr.Markdown(
        """
        # Dynamic Todo Workspace

        A simple interactive task management application.

        Add tasks, complete tasks, and remove tasks from the workspace.
        """
    )

    task_table = gr.Dataframe(
        value=get_task_table(),
        headers=[
            "Task ID",
            "Task",
            "Status"
        ],
        interactive=False,
        label="Current Tasks"
    )

    status_message = gr.Textbox(
        label="Status",
        interactive=False
    )

    gr.Markdown("## Add a Task")

    task_input = gr.Textbox(
        label="Task Description",
        placeholder="Enter a new task..."
    )

    add_button = gr.Button(
        "Add Task"
    )

    gr.Markdown("## Manage a Task")

    task_id_input = gr.Number(
        label="Task ID",
        precision=0
    )

    with gr.Row():

        complete_button = gr.Button(
            "Complete Task"
        )

        delete_button = gr.Button(
            "Delete Task"
        )

    add_button.click(
        fn=add_task,
        inputs=task_input,
        outputs=[
            task_table,
            status_message
        ]
    )

    complete_button.click(
        fn=complete_task,
        inputs=task_id_input,
        outputs=[
            task_table,
            status_message
        ]
    )

    delete_button.click(
        fn=delete_task,
        inputs=task_id_input,
        outputs=[
            task_table,
            status_message
        ]
    )

demo.launch()
```
